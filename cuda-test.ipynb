# Backend-agnostic test

```python
import os
STEPS = 50
```

```python
os.environ["KERAS_BACKEND"] = "torch" # 🔁 CHANGE THIS PER ENV: "tensorflow" | "torch" | "jax"
import torch
print("CUDA available:", torch.cuda.is_available())
print("GPU:", torch.cuda.get_device_name(0) if torch.cuda.is_available() else None)

import time
import keras
import numpy as np

print("Backend:", keras.backend.backend())

# -------------------------
# Benchmark configuration
# -------------------------
BATCH_SIZE = 1024
INPUT_DIM = 2048
HIDDEN_DIM = 4096
OUTPUT_DIM = 10
DTYPE = "float32"

# -------------------------
# Generate data
# -------------------------
x = np.random.randn(BATCH_SIZE, INPUT_DIM).astype(DTYPE)
y = np.random.randint(0, OUTPUT_DIM, size=(BATCH_SIZE,))

# -------------------------
# Build model
# -------------------------
model = keras.Sequential(
    [
        keras.layers.Dense(HIDDEN_DIM, activation="relu"),
        keras.layers.Dense(HIDDEN_DIM, activation="relu"),
        keras.layers.Dense(OUTPUT_DIM),
    ]
)

model.compile(
    optimizer="adam",
    loss=keras.losses.SparseCategoricalCrossentropy(from_logits=True),
)

# -------------------------
# Warm-up (important!)
# -------------------------
model.train_on_batch(x, y)

# -------------------------
# Timed run
# -------------------------
start = time.perf_counter()

for _ in range(STEPS):
    model.train_on_batch(x, y)

elapsed = time.perf_counter() - start

print(f"Total time ({STEPS} steps): {elapsed:.3f} s")
print(f"Time per step: {elapsed / STEPS:.4f} s")
```

\* *Assuming CPU execution based on the absence of GPU info in the output*

## Key Findings

- **GPU Acceleration**: PyTorch with CUDA on RTX 4070 is **8.6× faster** than PyTorch on CPU
- **Backend Comparison (CPU)**: JAX (6.972s) > TensorFlow (8.805s) > PyTorch (13.020s)
- **Fastest Overall**: PyTorch + GPU (1.509s total, 0.0302s per step)
- **CPU Performance**: JAX is the fastest CPU backend, ~1.9× faster than PyTorch CPU and ~1.3× faster than TensorFlow

The RTX 4070 GPU provides substantial acceleration, being **4.6× faster** than even the most optimized CPU backend (JAX).
